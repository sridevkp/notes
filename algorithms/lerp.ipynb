```python
import matplotlib.pyplot as plt
import numpy as np
from math import sqrt, pi
```

## Square

```python
x=np.linspace(0,1,10000)
y=x**2
plt.plot(x,y)
```

## Parabolic

```python
y=pow(4.0*x*(1.0-x),5)
plt.plot(x,y)
```

## Square Root

```python
y=np.sqrt(x)
plt.plot(x,y)
```

## Quadratic Ease out

```python
y=1.0 -(1.0-x) * (1.0-x)
plt.plot(x,y)
```

## Triangle

```python
y=1.0 - 2.0* abs(x-0.5)
plt.plot(x,y)
```

## Elastic Out 

```python
y=np.sin(-13.0 * (x+1.0) * pi/2) * pow(2.0, -10.0* x) + 1.0
plt.plot(x,y)
```

## Ease In Out


```python
y1 = x**2
y2 = 1.0 -(1.0-x) * (1.0-x)
y = y1 * (1.0 - x) + y2 * x
plt.plot(x,y)
```
