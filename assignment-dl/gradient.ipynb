```python
X = [.5, 2.5]
Y = [.2,  .9]
w, b = -1, -1
```

```python
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

def predict(weight, bias, x):
    return sigmoid(weight * x + bias)

def squared_error(y_pred, y_true):
    return (y_pred - y_true) ** 2

def d_loss_d_prediction(y_pred, y_true):
    return y_pred - y_true

def d_sigmoid(z):
    s = sigmoid(z)
    return s * (1.0 - s)

def d_loss_d_weight(weight, bias, x, y):
    z = weight * x + bias
    y_pred = sigmoid(z)
    return d_loss_d_prediction(y_pred, y) * d_sigmoid(z) * x

def d_loss_d_bias(weight, bias, x, y):
    z = weight * x + bias
    y_pred = sigmoid(z)
    return d_loss_d_prediction(y_pred, y) * d_sigmoid(z)

def train(weight, bias, X, Y, learning_rate=0.1, epochs=1000):
    for _ in range(epochs):
        grad_w = 0.0
        grad_b = 0.0

        for x, y in zip(X, Y):
            grad_w += d_loss_d_weight(weight, bias, x, y)
            grad_b += d_loss_d_bias(weight, bias, x, y)

        weight -= learning_rate * grad_w
        bias -= learning_rate * grad_b

    return weight, bias
```

```python
w, b = train(w, b, X, Y)
hist.append((w,b))
hist
```
